```python
import pandas as pd
from statsmodels.stats.contingency_tables import mcnemar
```

```python
path_a = "../output/avg_cal_lr_alpha_0.023.csv"
path_b = "../output/jump_cal_lr_alpha_0.023.csv"
```

```python
a = pd.read_csv(path_a)
b = pd.read_csv(path_b)

df = pd.DataFrame({
    "participant_id": a["participant_id"],
    "run": a["run"],
    "det_a": a["detected"],
    "det_b": b["detected"],
})

df
```

```python
summary = {}

for level in ["High", "Low"]:
    # Filter by level
    sub = df[df["run"] == level]

    # 2x2 paired contingency table (rows = method A detected, cols = method B detected)
    ct = (pd.crosstab(sub["det_a"], sub["det_b"])
            .reindex(index=[True, False], columns=[True, False], fill_value=0))

    # run McNemal test
    res = mcnemar(ct.to_numpy(), exact=True)
    summary[level] = {"n": len(sub), "mcnemar_p": res.pvalue}

summary = pd.DataFrame(summary).T
summary
```
